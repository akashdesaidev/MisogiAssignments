```python
summary= pd.DataFrame({"columns":df.columns,"dtype":df.dtypes.values,"missing":df.isnull().sum().values,"unique": df.nunique().values}) 
summary=summary.sort_values("missing",ascending=False)
```

```python
summary
```

```python
# Value counts & proportions
# For Pclass, Gender, and Embarked, show value counts and percentages (as a DataFrame).
cols=["Pclass","Sex","Embarked"]

for col in cols:
    values=df[col].value_counts(dropna=False)
    proportions=df[col].value_counts(normalize=True,dropna=False)*100

    result=pd.DataFrame({"values":values,"proportions":proportions.round(2)})
    print(result)

```

```python
female_firstclass_over_30=df[(df["Sex"]=="female")&(df["Pclass"]==1)&(df["Age"]>30)]
female_firstclass_over_30=female_firstclass_over_30.sort_values("Fare",ascending=False)
```

```python
print(df["Age"].mean().round(2))
print(df["Age"].median())
print(df["Age"].mode()[0])
```

```python
df.groupby("Pclass")["Fare"].mean()
```

```python
print("\nOverall survival rate:")
print(df["Survived"].mean().round(2))
```

```python
print("Survival rate by Gender:")
print(df.groupby("Sex")["Survived"].mean().round(2))
```

```python
print("Before filling missing:",df["Age"].isnull().sum())
df["Age_imputed"] = df.groupby(["Pclass", "Sex"])["Age"].transform(
    lambda x: x.fillna(x.median())
)
print("After filling missing:",df["Age_imputed"].isnull().sum())
```

```python
for group_keys, group_df in df.groupby(["Pclass", "Sex"]):
    print("Group:", group_keys)
    print(group_df.head(3))  # show first 3 rows from this group
    print("-" * 50)
```

```python
df["Title"] = df["Name"].str.extract(r",\s*([^\.]+)\.")

print("\nTitle counts:")
print(df["Title"].value_counts())
```

```python
df["family_size"]=df["SibSp"]+df["Parch"]+1
df["isAlone"]=df["family_size"]==1
print("Alone survived:", df.groupby("isAlone")["Survived"].mean()) 
```

```python
pivot = df.pivot_table(index="Pclass",columns="Sex",values="Survived",aggfunc="mean")
print(pivot)
```

```python
grouped = df.groupby(["Pclass", "Embarked"])[["Fare", "Age_imputed"]].mean()

print("\nGroupby - Avg Fare & Age_imputed by Pclass & Embarked:")
print(grouped)
```

```python
df["Cabin"]=df["Cabin"].fillna("Unknown")
df["CabinDeck"]=df["Cabin"].str[0]
print(df.head())
```
